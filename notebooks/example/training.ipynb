```python
from hydra_zen import launch

from yggdrasil.train import ZENSTORE, RunCfg, register_config, run
```

```python
register_config(RunCfg, ZENSTORE)
```

```python
import os
default_overrides = (
    "experiment=example-deep "
    "model.net.input_dim=128 "
    "paths.log_dir=./logs "
)
overrides = default_overrides.split()
```

```python
job = launch(RunCfg, run, version_base="1.3", overrides=list(overrides))
```

```python
import subprocess

assert job.cfg is not None, "Log directory must be specified in the configuration."

mlflow_dir = job.cfg["paths"]["log_dir"] + "/mlflow/mlruns"
print(f"MLflow runs are stored in: {mlflow_dir}")

try:
    subprocess.run(["mlflow", "ui", "--backend-store-uri", f"file:{mlflow_dir}"])
except KeyboardInterrupt:
    print("MLflow UI stopped by user.")
```
